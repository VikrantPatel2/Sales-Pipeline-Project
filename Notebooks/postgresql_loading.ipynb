```python
import pandas as pd
from sqlalchemy import create_engine,text

df = pd.read_csv("cleaned_sales.csv")

engine = create_engine(
    "postgresql+psycopg2://postgres:YourPassword@localhost:5432/sales_automation"
)
df.to_sql(
    "sales",
    engine,
    if_exists="replace",
    index=False
)

print("Data loaded successfully!")
```
